# 1. 표준 폴더 구조 확인·생성

```python
# 표준 하위 폴더가 없으면 만들고, 있으면 그대로 둔다(exist_ok=True)
for name in ['data', 'notebooks', 'docs']:
    d = ROOT / name
    d.mkdir(parents=True, exist_ok=True)
    n_items = len([p for p in d.glob('*') if not p.name.startswith('.')])
    print(f'{name:10s} 항목 {n_items}개')

# 루트 구성을 한눈에 출력
print('\n[gen-ai 루트 구성]')
for p in sorted(ROOT.iterdir()):
    if p.name.startswith('.'):
        continue
    print(('[D] ' if p.is_dir() else '[F] ') + p.name)
```
