# 런타임 점검과 requirements


### 1. 지금 런타임 확인 (코랩 여부 / 파이썬 / GPU)

```python
import sys, platform

# 'google.colab' 모듈은 코랩에서만 import 된다 → 환경 판별
try:
    import google.colab  # noqa
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

print('실행 환경 :', 'Google Colab' if IN_COLAB else '로컬/기타')
print('파이썬    :', sys.version.split()[0])   # 코랩 3.12.x / 로컬 genai 3.11.x
print('플랫폼    :', platform.platform())
```

### 2. 핵심 패키지 설치 여부 점검

```python
import importlib.util
def installed(mod):
    try:
        return importlib.util.find_spec(mod) is not None
    except ModuleNotFoundError:
        return False

for mod in ['google.genai', 'pandas', 'transformers', 'chromadb']:
    print(f'{mod:16s} {"O" if installed(mod) else "X (필요 시점에 설치)"}')
```

### 3. requirements.txt 다루기

```python
# 드라이브 마운트 — 처음 실행 시 '허용' 팝업을 클릭
from google.colab import drive
drive.mount('/content/drive')

# gen-ai 루트에 샘플 requirements 를 만들어 읽어본다
from pathlib import Path
ROOT = Path('/content/drive/MyDrive/kt cloud tech up/gen-ai')
req = ROOT / 'requirements.txt'
req.write_text('google-genai\npython-dotenv\npandas==2.2.3\n', encoding='utf-8')

lines = [ln.strip() for ln in req.read_text(encoding='utf-8').splitlines()
         if ln.strip() and not ln.strip().startswith('#')]
print('선언된 패키지:', lines)
print('설치 명령: pip install -r requirements.txt')
```
